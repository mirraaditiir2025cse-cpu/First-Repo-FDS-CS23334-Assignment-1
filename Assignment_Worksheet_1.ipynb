```python
#Creating a DataFrame
import pandas as pd
data={
    'Name':['Alice', 'Bob', 'Charlie', 'David'],
    'Age':[25, 30, 35, 40],
    'City':['New York', 'Los Angeles', 'Chicago', 'Houston']
}
print(data)
```

```python
#display the DataFrame
df=pd.DataFrame(data)
print(df)
```

```python
#display the rows where age is above than 30
filter = df[df['Age'] > 30]
print(filter)
```

```python
#Adding a new column
df['Salary']=[50000,60000,70000,80000]
print(df)
```

```python
#updating the common values
import pandas as pd
data={
    'Name':['Alice','Bob','Charlie','David'],
    'Age':[25,30,35,40],
    'City':['New York', 'Los Angeles', 'Chicago', 'Houston'],
    'Salary':[50000,60000,70000,80000]
}

df=pd.DataFrame(data)

df["Salary"]=df["Salary"]*10/100
print(df)
```

```python
#deleting a Column from df
df=df.drop("City",axis=1)
print(df)
```

```python
#Renaming a column in df
df=df.rename(columns={'Age':'Years'})
print(df)
```

```python
#arranging the values in descending order
df=df.sort_values(by='Salary',ascending=False)
print(df)
```

```python
#Creating a new DataFrame
data2={
    'Department':['HR','IT','HR','IT'],
    'Employee':['Alice', 'Bob', 'Charlie', 'David'],
    'Salary':[50000,60000,55000,65000]
}

df2=pd.DataFrame(data2)
print(df2)

grouped=df2.groupby("Department")

average_salary=grouped["Salary"].mean()
print("\nAverage salary from the Department:")
print(average_salary)
```

```python
#merging two dataFrame
df1=pd.DataFrame({
    'ID':[1,2,3],
    'Name':['Alice', 'Bob', 'Charlie']
})
df2=pd.DataFrame({
    'ID':[1,2,4],
    'Score':[85,90,88]
})
merged_df=pd.merge(df1,df2, on='ID')
print(merged_df)
```

```python
#Handling missing values
data={
    'Name':['Alice', None, 'Charlie'],
    'Age':[25,30,None],
    'City':[None, 'Los Angeles', 'Chicago']
}
df=pd.DataFrame(data)
df["Name"]=df["Name"].fillna("NULL")
df["Age"]=df["Age"].fillna("NULL")
df["City"]=df["City"].fillna("NULL")

print(df)
```
